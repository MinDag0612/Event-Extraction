```python
from pathlib import Path
import sys
import json

ROOT = next(
    path for path in [Path.cwd(), *Path.cwd().parents]
    if (path / "src/adapter/PHEE_adapter.py").exists()
)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
```

```python
from src.adapter.PHEE_adapter import PHEEAdapter
```

```python
with open(ROOT / "data/raw/PHEE/train.json", "r", encoding="utf-8") as file:
    data = json.loads(next(file))
```

```python
adapter = PHEEAdapter()
sample = adapter.adapt(data)

print(json.dumps(json.loads(sample.to_json()), ensure_ascii=False, indent=2))
```

```python
# Thống kê các split (chạy riêng khi cần kiểm tra toàn bộ dữ liệu).
for split in ["train", "dev", "test"]:
    sentence_count = 0
    event_count = 0
    with open(ROOT / f"data/raw/PHEE/{split}.json", "r", encoding="utf-8") as file:
        for line in file:
            if not line.strip():
                continue
            record = adapter.adapt(json.loads(line))
            sentence_count += 1
            event_count += len(record.events)
    print(split, "sentences:", sentence_count, "events:", event_count)
```
