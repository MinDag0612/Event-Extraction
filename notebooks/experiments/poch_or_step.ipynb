```python
import json
from pathlib import Path

root = Path("D:/Event-Extraction/model/Qwen3-4B/checkpoint-BKEE")

checkpoints = []

for path in root.glob("checkpoint-*"):
    state_file = path / "trainer_state.json"

    if not state_file.exists():
        continue

    with open(state_file, "r") as f:
        state = json.load(f)

    checkpoints.append({
        "name": path.name,
        "step": state["global_step"],
        "epoch": state["epoch"],
    })

for ckpt in sorted(checkpoints, key=lambda x: x["step"]):
    print(
        f'{ckpt["name"]:20} | '
        f'Step: {ckpt["step"]:6} | '
        f'Epoch: {ckpt["epoch"]:.4f}'
    )
```
